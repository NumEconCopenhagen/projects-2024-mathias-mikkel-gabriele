# Real Business Cycles Theory

The standard RBC framework has a representative households that maximize the expectation of their utility at time $t=0$ by choosing their optimal consumption $\{c_t \}_{t=0}^{T}$ and leisure $\{l_t \}_{t=0}^{T}$

$$E_0 \left[\sum_{t=0}^{T}{\beta^t u(c_t,l_t)} \right]$$



The households are constrained in multiple ways:
- Hours endowment:          $n_t+l_t=1$
- Ressource constraint:     $y_t = c_t+i_t$

In order to produce income, the household can supply labor and invest in capital. Capital follow the following law of motion.
$$k_{t+1} = i_t + (1-\delta)k_t$$

The production is a neoclassical production function
$$y_t=A_t F(k_t,n_t)$$

why the real rate of return and the real wage can be computed as respectively the FOC wrt. $k$ and $n$

$$\Rightarrow r_t = F'_k(k_t,n_t)$$
$$\Rightarrow w_t = F'_n(k_t,n_t)$$

The technology follows a stochastic AR(1)-process with $\epsilon_t \sim i.i.d.(0,\sigma^2_{\epsilon})$
$$\log(A_t) = (1-\rho) \mu_A +\rho \log(A_{t-1})+\epsilon_t$$

The households face the following problem with the hours endowments inserted for $l_t$, $MPK$ inserted for $r_t$ and $MPL$ inserted for $w_t$
$$\max_{\{c_t,k_{t+1},n_t \}_{t=0}^{T}} {E_0 \left[ \sum_{t=0}^{T} {\beta^t u(c_t,1-n_t)} \right]}$$

$$s.t.  k_{t+1} = w_t n_t + r_t k_t - c_t +(1-\delta)k_t = F'_n (k_t,n_t) n_t + F'_n(k_t,n_t) k_t - c_t +(1-\delta)k_t$$

From this, the Langranian can be formulated:
$$\mathcal{L} = E_0 \left[ \sum_{t=0}^{T} {\beta^t \left[u(c_t,1-n_t) +\lambda_t ( F'_n (k_t,n_t) n_t + F'_n(k_t,n_t) k_t - c_t +(1-\delta)k_t - k_{t+1} )\right]}  \right]$$

And the FOCS determining optimal behavior can be derived to be:
$$u'_c(c_t,1-n_t) = \beta E_t \left[ u'_c(c_{t+1},1-n_{t+1}) (1-A_{t+1} F'_k (k_{t+1},n_{t+1})-\delta) \right]$$

$$u'_n (c_t,1-n_t) = u'_c(c_t,1-n_t) A_t F'_n (k_t,n_t)$$

With the law of motion for capital and the stochastic technology proces, the equation system is complete

$$k_{t+1} = i_t + (1-\delta)k_t$$
$$\log(A_t) = (1-\rho) \mu_A +\rho \log(A_{t-1})+\epsilon_t$$
